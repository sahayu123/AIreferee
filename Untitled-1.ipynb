```python
import json
import shutil
import pandas as pd
from pathlib import Path

# Base dataset paths
dataset_path = Path("/Users/rishideshpande/Downloads/data/soccernet/mvfouls")
labels_path = dataset_path / "processed_frames" / "test" / "labels.csv"
train_annotations = dataset_path / "train" / "annotations.json"

# NEW output directory: Placed next to 'mvfouls' inside 'soccernet'
clean_output_dir = dataset_path.parent / "processed_frames_no_handball" / "test"
clean_output_dir.mkdir(parents=True, exist_ok=True)

# 1. Load annotations to identify Handball action IDs
with open(train_annotations, "r") as f:
    annotations = json.load(f)["Actions"]

handball_action_ids = set()
for action_id, info in annotations.items():
    if info.get("Offence") == "Handball" or info.get("Offence_Type") == "Handball":
        handball_action_ids.add(str(action_id))

print(f"Found {len(handball_action_ids)} Handball action IDs to exclude.")

# 2. Read existing CSV and filter out handball rows
df = pd.read_csv(labels_path)
df["action_id"] = df["action_id"].astype(str)
filtered_df = df[~df["action_id"].isin(handball_action_ids)].copy()

# 3. Copy image files to the new directory outside mvfouls
new_image_paths = []

for idx, row in filtered_df.iterrows():
    old_path = Path(row["image_path"])
    
    # Preserves relative subfolder structure: action_X/clip_Y/frame_Z.jpg
    relative_structure = Path(*old_path.parts[-3:]) 
    new_path = clean_output_dir / relative_structure
    
    # Create subdirectories if needed
    new_path.parent.mkdir(parents=True, exist_ok=True)
    
    # Copy file if it hasn't been copied yet
    if old_path.exists() and not new_path.exists():
        shutil.copy2(old_path, new_path)
        
    new_image_paths.append(str(new_path))

# 4. Save updated DataFrame with new file paths inside the new directory
filtered_df["image_path"] = new_image_paths
filtered_df.to_csv(clean_output_dir / "labels.csv", index=False)

print("\n--- Export Complete ---")
print(f"Clean images copied to: {clean_output_dir}")
print(f"Total valid frames copied: {len(filtered_df)}")
```
